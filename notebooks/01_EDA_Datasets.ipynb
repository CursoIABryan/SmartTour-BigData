```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Cargar los datasets
municipios = pd.read_csv(
    "../datasets/municipios/municipios.csv",
    sep=";",
    encoding="cp1252",
    decimal=","
)

hoteles = pd.read_csv(
    "../datasets/hoteles/hoteles.csv",
    sep=";",
    encoding="latin1",
    decimal=","
)

monumentos = pd.read_csv(
    "../datasets/monumentos/monumentos.csv",
    sep=";",
    encoding="latin1",
    decimal=","
)

oficinas_turismo = pd.read_csv(
    "../datasets/oficinas_turismo/oficinas_turismo.csv",
    sep=";",
    encoding="latin1",
    decimal=","
)

# 2. Agrupar los datasets en un diccionario
datasets = {
    "Municipios": municipios,
    # "Hoteles": hoteles,
    # "Monumentos": monumentos,
    # "Oficinas de turismo": oficinas_turismo
}

# 3. Analizar cada dataset
for nombre, df in datasets.items():

    print(f"\n========== {nombre.upper()} ==========")

    # Número de filas y columnas
    print("\n1. Número de filas: (municipios.shape)")
    print(df.shape)

    # Información y tipos de columnas
    print("\n2. Tipos de columnas: (municipios.info())")
    df.info()

    # Primeros 5 registros
    print("\n3. Primeros registros: (municipios.head())")
    display(df.head())

    # Estadísticas descriptivas
    print("\n4. Estadísticas: (municipios.describe())")
    display(df.describe())

    # Valores nulos por columna
    print("\n5. Valores nulos: (municipios.isnull().sum())")
    print(df.isnull().sum())

    # Número de registros duplicados
    print("\n6. Valores duplicados: (municipios.duplicated().sum())")
    print(df.duplicated().sum())

    # Nombres de las columnas
    print("\n7. Nombres de las columnas: (municipios.columns)")
    print(df.columns)
```
